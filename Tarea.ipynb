```python
#Importamos librerías requeridas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Carga desde un archivo .xlsx sin indice
df= pd.read_excel('Citas_Digital.xlsx')
df
```

```python
#Verificamos información del DataFrame
df.info()
```

```python
#Paso 1 - Filtro por filas
filtro1 = df.iloc[0:673, :]
filtro1.info()
```

```python
filtro2 = filtro1.iloc[:, [0,1,2,4,5,6,7,8,9]]
filtro2.info()
```

```python
#Paso 3 - Convertir Potencial de compra a float
filtro2["Potencial de compra"] = filtro2["Potencial de compra"].astype(float)
filtro2.info()
```

```python
#Paso 4 - Crear CSV General
filtro2.to_csv("Citas_Digital_General.csv", index=False)
```
